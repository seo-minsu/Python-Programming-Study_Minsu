Data Filtering

* Indexing
  - .loc[]: 조건문이나 label을 통해 행/열 선택
    
    df.loc[2, 'Grade']
  - .iloc[]: index를 기반으로 행/열 선택

* 특정 컬럼만 필터링
  - df[['Name', 'Grade']]

* 조건으로 필터링
  - df[df['age']>20]
  - df[df['grade']=="B"]

* 특정 행/열 제거
  - .drop(): 괄호 안에 열 이름이나 행/열의 index 작성
  - axis가 1이면 열을 제거, 0이면 행을 제거
   
   df.drop(['Age'], axis=1)

* 중복치 제거
  - .drop_duplicates(): 데이터프레임 내의 중복 행 제거
  - 보통은 가장 마지막에 진행

  df.drop_duplicates()

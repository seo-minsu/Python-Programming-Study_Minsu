```python
data = [
	{"Name":"Joe",    "Age":20, "Score":85, "Grade":"A", "Hobby":"Swimming"},
	{"Name":"Nat",    "Age":21, "Score":77, "Grade":"B", "Hobby":"Reading"},
	{"Name":"Harry",  "Age":19, "Score":91, "Grade":"A", "Hobby":"Music"},
	{"Name":"Joe",    "Age":20, "Score":85, "Grade":"A", "Hobby":"Swimming"},
	{"Name":"Monica", "Age":22, "Score":60, "Grade":"B", "Hobby":"Dancing"}
]
```

```python
import pandas as pd

df = pd.DataFrame(data)
print(df)
```

```python
print(df.loc[2,'Grade'])
print(df.iloc[3,2])
```

```python
target_columns = ['Name', 'Grade']

filtered_df = df[target_columns]

print(filtered_df)
```

```python
print(df[df['Age']>20])
```

```python
adult_df = df[df['Grade']=='A']

print(adult_df)
```

```python
dropped_df = df.drop(['Age'], axis=1)
print(dropped_df)
```

```python
no_duplicate_df = df.drop_duplicates()
print(no_duplicate_df)
```

```python
from sklearn.datasets import load_iris # scikit-learn의 샘플 데이터 로드를 위해 import

iris = load_iris() # sample data load
```

```python
import pandas as pd

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target
df['target'] = df['target'].map({0:"setosa", 1:"versicolor", 2:"virginica"})
```

```python
print(df)
```
